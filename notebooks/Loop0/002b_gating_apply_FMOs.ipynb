# Loop0, stage 02b: apply the gating strategy to the controls

The controls have to pass exactly the same three gates as the screen, or the
background they measure would come from a different set of cells.

## Inputs

| file | produced by | holds |
|---|---|---|
| `Loop0_FMOs.h5ad` | stage 01b | every control cell |
| `gates.npz` | stage 01 | the three gate polygons |

## Outputs

| file | holds |
|---|---|
| `Loop0_FMOs_Gated.h5ad` | the control cells that passed all three gates |

Run the cells from top to bottom.

```python
import matplotlib.pyplot as plt
import pytometry as pm
import scanpy as sc

from labcompass import apply_saved_gates

IN_FMO = "Loop0_FMOs.h5ad"
GATE_FILE = "gates.npz"
OUT_FMO_GATED = "Loop0_FMOs_Gated.h5ad"
```

```python
adata_all = sc.read(IN_FMO)
before = adata_all.n_obs
adata_all, masks = apply_saved_gates(adata_all, GATE_FILE, mode="all")
print(f"{before:,} cells -> {adata_all.n_obs:,} cells "
      f"({100 * adata_all.n_obs / before:.1f} per cent)")
```

```python
pm.pl.scatter_density(
    adata_all, x="FSC-A :: FSC - Area", y="7-AAD",
    x_lim=[0, 1_500_000], y_lim=[-10_000, 10_000],
)
```

```python
plt.close("all")
```

## Cells kept per control

```python
counts = adata_all.obs["fmo"].value_counts()
ax = counts.plot.bar(figsize=(8, 4))
ax.set_ylabel("cells kept")
plt.tight_layout()
plt.show()
counts
```

```python
adata_all.write(OUT_FMO_GATED)
print(f"wrote {OUT_FMO_GATED}: {adata_all.n_obs:,} cells")
```
